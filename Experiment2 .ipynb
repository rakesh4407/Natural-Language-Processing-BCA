```python
from nltk.tokenize import word_tokenize, sent_tokenize

import nltk

nltk.download('punkt')
```

```python
text = "Rakesh is Playing Chess. He lives in Hostel, Haryana. He likes Finance"

words = word_tokenize(text)
sentences = sent_tokenize(text)

print(words)
print(sentences)
print("And the length is: ", len(words))
```

```python
# Removing Puntuation

import string
from nltk.tokenize import word_tokenize

text = "Hello Mr. Smith, how are you doing today? The weather is great, and Python is awesome. The sky is pinkish-blue. You shouldn't eat cardboard."

tokens = word_tokenize(text)

words = []

for word in tokens:
    if word not in string.punctuation:
        words.append(word)

print("\n \n NEW: ", words) 
```

```python
# Stop word Removal

from nltk.corpus import stopwords
nltk.download('stopwords')
text =" he is a good student, and a good tutor also. But he can't write books."
tokens = word_tokenize(text)

example = []
for i in tokens:
    if i.lower() not in stopwords.words('english'):
        example.append(i)
example
```
